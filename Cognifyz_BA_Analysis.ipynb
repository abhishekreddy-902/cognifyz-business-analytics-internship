```python

# ========================
# TASK 2: GENDER DISTRIBUTION
# ========================

# value_counts() counts how many of each category
gender_counts = df['gender'].value_counts()

print("Gender Count:")
print(gender_counts)

# normalize=True gives percentage instead of count
# mul(100) converts 0.625 to 62.5
# round(1) rounds to 1 decimal place
print("\nGender Percentage:")
print(df['gender'].value_counts(normalize=True).mul(100).round(1))
```

```python
# figsize=(12,5) means width=12 inches, height=5 inches
# subplots(1,2) means 1 row, 2 columns = 2 charts side by side
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# --- CHART 1: Pie Chart (axes[0] means first chart) ---
colors = ['#3498db', '#e74c3c']   # blue for male, red for female

axes[0].pie(
    gender_counts.values,          # the numbers (25, 15)
    labels=gender_counts.index,    # the labels (Male, Female)
    autopct='%1.1f%%',             # show percentage like 62.5%
    colors=colors,
    startangle=90                  # start from top
)
axes[0].set_title('Gender - Pie Chart', fontsize=13, fontweight='bold')

# --- CHART 2: Bar Chart (axes[1] means second chart) ---
bars = axes[1].bar(
    gender_counts.index,           # x-axis: Male, Female
    gender_counts.values,          # y-axis: 25, 15
    color=colors,
    edgecolor='black',
    width=0.4
)

# This loop adds the number on top of each bar
for bar in bars:
    height = bar.get_height()      # get the bar's height value
    axes[1].text(
        bar.get_x() + bar.get_width()/2,   # x position = center of bar
        height + 0.2,                       # y position = just above bar
        str(int(height)),                   # the number to display
        ha='center',                        # center aligned
        fontweight='bold',
        fontsize=12
    )

axes[1].set_title('Gender - Bar Chart', fontsize=13, fontweight='bold')
axes[1].set_ylabel('Number of Investors')
axes[1].set_ylim(0, 30)           # y-axis goes from 0 to 30

# Main title for both charts
plt.suptitle('Task 2: Gender Distribution', fontsize=15, fontweight='bold')
plt.tight_layout()                 # adjusts spacing so nothing overlaps

# Save the chart as a PNG image file
plt.savefig('Task2_Gender_Distribution.png', dpi=150, bbox_inches='tight')
plt.show()                         # display the chart in notebook
print("✅ Chart saved as Task2_Gender_Distribution.png")
```

```python
# ========================
# TASK 3: DESCRIPTIVE STATISTICS
# ========================

# These are all the numerical columns in our dataset
num_cols = ['age', 'Mutual_Funds', 'Equity_Market', 'Debentures',
            'Government_Bonds', 'Fixed_Deposits', 'PPF', 'Gold']

# .agg() applies multiple functions at once
# mean = average, median = middle value, std = spread of data
# min = smallest value, max = largest value
stats_table = df[num_cols].agg(['mean', 'median', 'std', 'min', 'max']).round(2)

print("=== Descriptive Statistics ===")
print(stats_table)
```

```python
# FINDING:
# Average investor age = 27.8 years (young investors!)
# Gold has highest average score (5.98) among investments
# PPF has lowest average score (2.02)
```

```python
# ========================
# TASK 4: MOST PREFERRED INVESTMENT AVENUE
# ========================

# Count how many investors prefer each avenue
avenue_counts = df['Avenue'].value_counts()

print("Investment Avenue Counts:")
print(avenue_counts)

# index[0] = first item (most common)
# values[0] = its count number
print(f"\n✅ Most Preferred: {avenue_counts.index[0]}")
print(f"   Number of investors: {avenue_counts.values[0]} out of 40")

# --- Create Bar Chart ---
plt.figure(figsize=(10, 5))       # single chart, width=10, height=5

colors = ['#2ecc71', '#3498db', '#e74c3c', '#f39c12']

bars = plt.bar(
    avenue_counts.index,           # x-axis labels
    avenue_counts.values,          # bar heights
    color=colors,
    edgecolor='black',
    width=0.5
)

# Add numbers on top of each bar
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height + 0.2,
        str(int(height)),
        ha='center', fontweight='bold', fontsize=12
    )

plt.title('Task 4: Most Preferred Investment Avenue',
          fontsize=14, fontweight='bold')
plt.ylabel('Number of Investors')
plt.xlabel('Investment Avenue')
plt.ylim(0, max(avenue_counts.values) + 4)
plt.tight_layout()
plt.savefig('Task4_Investment_Avenue.png', dpi=150, bbox_inches='tight')
plt.show()
print("✅ Chart saved!")
```

```python
# ========================
# TASK 5: REASONS FOR INVESTMENT
# ========================

print("=== TASK 5: REASONS FOR INVESTMENT ===\n")

# We have 4 different reason columns
# We loop through each one and print the counts

reason_cols = {
    'Reason_Equity': 'Equity Investment Reasons',
    'Reason_Mutual': 'Mutual Fund Reasons',
    'Reason_Bonds': 'Bond Investment Reasons',
    'Reason_FD':    'Fixed Deposit Reasons'
}

# .items() gives you both key and value from the dictionary
for column_name, display_title in reason_cols.items():
    print(f"--- {display_title} ---")
    print(df[column_name].value_counts())
    print()   # empty line for spacing
```

```python
# ========================
# TASK 6: SAVINGS OBJECTIVES
# ========================
print("=== TASK 6: SAVINGS OBJECTIVES ===")
savings = df['What are your savings objectives?'].value_counts()
print(savings)
print(f"\nTop savings goal: {savings.index[0]} ({savings.values[0]} investors)")
```

```python
# ========================
# TASK 8: INVESTMENT DURATION
# ========================
print("=== TASK 8: INVESTMENT DURATION ===")
duration = df['Duration'].value_counts()
print(duration)

# Map text values to numbers to calculate average
# Less than 1 year → 0.5, 1-3 years → 2, etc.
duration_map = {
    'Less than 1 year': 0.5,
    '1-3 years': 2,
    '3-5 years': 4,
    'More than 5 years': 6
}

# .map() replaces each text value with the number from our dictionary
df['Duration_Numeric'] = df['Duration'].map(duration_map)

avg = df['Duration_Numeric'].mean().round(2)
print(f"\nApproximate Average Investment Duration: {avg} years")
```

```python
# ========================
# TASK 9: EXPECTATIONS FROM INVESTMENTS
# ========================
print("=== TASK 9: EXPECTATIONS ===")
expect = df['Expect'].value_counts()
print(expect)
print(f"\nTop expectation: {expect.index[0]} ({expect.values[0]} investors)")
```

```python
# ========================
# TASK 10: CORRELATION ANALYSIS
# ========================

num_cols = ['age', 'Mutual_Funds', 'Equity_Market', 'Debentures',
            'Government_Bonds', 'Fixed_Deposits', 'PPF', 'Gold']

# .corr() calculates correlation between all numerical columns
corr_matrix = df[num_cols].corr().round(2)

print("Correlation Matrix:")
print(corr_matrix)

# --- Heatmap ---
plt.figure(figsize=(10, 7))

# sns.heatmap draws the colored grid
# annot=True = show the numbers inside each box
# fmt='.2f' = show 2 decimal places
# cmap='RdYlGn' = Red(negative) Yellow(zero) Green(positive) colors
sns.heatmap(
    corr_matrix,
    annot=True,
    fmt='.2f',
    cmap='RdYlGn',
    linewidths=0.5,
    linecolor='white',
    vmin=-1, vmax=1,
    annot_kws={'size': 10}
)

plt.title('Task 10: Correlation Analysis Heatmap',
          fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('Task10_Correlation_Heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
print("✅ Heatmap saved!")
```
